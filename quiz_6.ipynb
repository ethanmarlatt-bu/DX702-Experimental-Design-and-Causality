### Imports

```python
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors
```

### Load Datasets

```python
df = pd.read_csv('datasets/homework_6.1.csv')
```

### Q1

```python
treated = df[df["X"] == 1].reset_index(drop=True)
untreated = df[df["X"] == 0].reset_index(drop=True)

Z_treated = treated[["Z"]].values
Z_untreated = untreated[["Z"]].values

nn_for_treated = NearestNeighbors(n_neighbors=1).fit(Z_untreated)
dist_t, idx_t = nn_for_treated.kneighbors(Z_treated)
treated_effects = treated["Y"].values - untreated["Y"].values[idx_t.flatten()]

nn_for_untreated = NearestNeighbors(n_neighbors=1).fit(Z_treated)
dist_u, idx_u = nn_for_untreated.kneighbors(Z_untreated)
untreated_effects = treated["Y"].values[idx_u.flatten()] - untreated["Y"].values

all_effects = np.concatenate([treated_effects, untreated_effects])

ATE = all_effects.mean()
print(f"ATE = {ATE:.4f}\n")

ATT = treated_effects.mean()
print(f"ATT = {ATT:.4f}\n")

ATU = untreated_effects.mean()
print(f"ATU = {ATU:.4f}\n")
```

```python
opt_idx = np.argmax(untreated_effects)
optimal_effect = untreated_effects[opt_idx]
matched_treated_idx = idx_u.flatten()[opt_idx]

print(f"Optimal effect = {optimal_effect:.4f}")

print("\nUntreated unit:\n", untreated.iloc[opt_idx])
print("\nMatched treated counterfactual:\n", treated.iloc[matched_treated_idx])
```
