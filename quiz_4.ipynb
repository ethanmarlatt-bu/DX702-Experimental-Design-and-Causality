### Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import statsmodels.api as sm
```

### Datasets

```python
df_1 = pd.read_csv('./datasets/homework_4.1.csv')

df_2_a = pd.read_csv('./datasets/homework_4.2.a.csv')
df_2_b = pd.read_csv('./datasets/homework_4.2.b.csv')
```

## Q1

```python
df_1.head()
```

```python
y_avg_instrument = df_1.loc[df_1['Z'] == 1, 'Y'].mean()
y_avg = df_1.loc[df_1['Z'] == 0, 'Y'].mean()
y_diff_instrument = y_avg_instrument - y_avg

x_avg_instrument = df_1.loc[df_1['Z'] == 1, 'X'].mean()
x_avg = df_1.loc[df_1['Z'] == 0, 'X'].mean()
x_diff_instrument = x_avg_instrument - x_avg

print(f"{y_diff_instrument / x_diff_instrument:.1f}")
```

## Q2

```python
n_bins = 25
df_1['W_bin'] = pd.cut(df_1['W'], bins=n_bins)

ratios = []
for wbin, g in df_1.groupby('W_bin', observed=True):
    z1 = g[g['Z'] == 1]
    z0 = g[g['Z'] == 0]

    if len(z1) == 0 or len(z0) == 0:
        continue

    y_diff = z1['Y'].mean() - z0['Y'].mean()
    x_diff = z1['X'].mean() - z0['X'].mean()

    ratios.append(y_diff / x_diff)

print(f"{np.mean(ratios):.1f}")
```

```python
cutoff = 80

print('4.2.a')
print('-------')

df_2_a['test_minus_cutoff'] = df_2_a.X - cutoff
df_2_a['takes_course'] = (df_2_a.X >= cutoff).astype(int)
df_2_a['interaction'] = df_2_a['test_minus_cutoff'] * df_2_a['takes_course']

X = sm.add_constant(df_2_a[['test_minus_cutoff', 'takes_course', 'interaction']])
model = sm.OLS(df_2_a.Y, X).fit()

b0 = model.params['const']
b1 = model.params['test_minus_cutoff']
b2 = model.params['takes_course']
b3 = model.params['interaction']
p2 = model.pvalues['takes_course']

print(f"beta_0 (intercept) = {b0:.4f}")
print(f"beta_1 (slope) = {b1:.4f}")
print(f"beta_2 (course effect) = {b2:.4f} (p = {p2:.4f})")
print(f"beta_3 (slope change) = {b3:.4f}")
```

```python
print('4.2.b')
print('-------')

df_2_b['test_minus_cutoff'] = df_2_b.X2 - cutoff
df_2_b['takes_course'] = (df_2_b.X2 >= cutoff).astype(int)
df_2_b['interaction'] = df_2_b['test_minus_cutoff'] * df_2_b['takes_course']

X = sm.add_constant(df_2_b[['test_minus_cutoff', 'takes_course', 'interaction']])
model = sm.OLS(df_2_b.Y2, X).fit()

b0 = model.params['const']
b1 = model.params['test_minus_cutoff']
b2 = model.params['takes_course']
b3 = model.params['interaction']
p2 = model.pvalues['takes_course']

print(f"beta_0 (intercept) = {b0:.4f}")
print(f"beta_1 (slope) = {b1:.4f}")
print(f"beta_2 (course effect) = {b2:.4f} (p = {p2:.4f})")
print(f"beta_3 (slope change) = {b3:.4f}")
```
