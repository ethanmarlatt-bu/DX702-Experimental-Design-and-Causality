### Q4

```python
def make_error(corr_const, num):
    err = list()
    prev = np.random.normal(0, 1) 
    for n in range(num): 
        prev = corr_const * prev + (1 - corr_const) * np.random.normal(0, 1) 
        err.append(prev) 
    return np.array(err) 

for cc in [0.2, 0.5, 0.8]:
    coefs = []
    ses = []
    
    for trial in range(1000):
        err_X = make_error(cc, 1000)
        X = err_X.copy()
        
        err_Y = make_error(cc, 1000)
        beta0_true, beta1_true = 2, 3
        Y = beta0_true + beta1_true * X + err_Y
        
        X_design = sm.add_constant(X)
        model = sm.OLS(Y, X_design).fit()
        
        coefs.append(model.params[1])
        ses.append(model.bse[1])
    
    print(f"ratio = {np.std(coefs)/np.mean(ses):.4f}")
```
