# Quiz 5

### Imports

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

### Given Script

```python
num = 100000 
 
difficulty = np.random.uniform(0, 1, (num,)) 
 
speed = np.maximum(np.random.normal(15, 5, (num, )) - difficulty * 10, 0) 
 
accident = np.minimum(np.maximum(0.03 * speed + 0.4 * difficulty + np.random.normal(0, 0.3, (num,)), 0), 1) 
 
df = pd.DataFrame({'difficulty': difficulty, 'speed': speed, 'accident': accident}) 
```

## Q1

```python
n_experiments = 5000
samples_per_experiment = 100000 

coefs = []
for _ in range(n_experiments):
    X = df[['difficulty']].values
    Y = df['speed'].values
    model = LinearRegression().fit(X, Y)
    coefs.append(model.coef_[0])

coefs = np.array(coefs)
print(f"avg coefficient (X): {coefs.mean():.4f}")
print(f"Std: {coefs.std():.4f}")
```

## Q2

```python
coefs_X = []
coefs_Z = []
for _ in range(n_experiments):
    X = df[['difficulty', 'accident']].values
    Y = df['speed'].values
    model = LinearRegression().fit(X, Y)
    coefs_X.append(model.coef_[0])
    coefs_Z.append(model.coef_[1])

coefs_X = np.array(coefs_X)
coefs_Z = np.array(coefs_Z)
print(f"avg coef of X: {coefs_X.mean():.4f}")
print(f"avg coef of Z: {coefs_Z.mean():.4f}")
```
