### imports

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
```

### datasets

```python
df_1 = pd.read_csv('datasets/homework_10.1.csv')
```

### Q1

```python
def get_fixed_effect_time(t):
    df_n = df_1[df_1['time'] == t]
    model = smf.ols('y ~ X', data=df_n).fit()
    return model.params['Intercept']

results = []
for i in range(0, 12):
    results.append(get_fixed_effect(i))

for i, result in enumerate(results):
    print(f"{i}: {result:.2f}")
```

### Q2

```python
def get_fixed_effect_city(c):
    df_c = df_1[df_1['city'] == c]
    model = smf.ols('y ~ X', data=df_c).fit()
    return model.params['Intercept']

results = []
for i in range(0, 10):
    results.append(get_fixed_effect(i))

for i, result in enumerate(results):
    print(f"{i}: {result:.2f}")
```

### Q3

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

np.random.seed(42)

num = 10000

X = np.clip(np.random.normal(3, 1, (num,)), 0.01, 100)
Z = np.clip(np.random.normal(3, 1, (num,)), 0.01, 100)
Y = np.log(X + Z) + np.random.normal(0, 1, (num,))

df = pd.DataFrame({'X': X, 'Z': Z})
df['expY'] = np.exp(Y)

model = smf.ols('expY ~ X + Z', data=df).fit()
print(model.params)
```

### Q4

```python
num = 10000

Z = np.random.normal(0, 1, (num,))
X = Z + np.random.normal(0, 1, (num,))
Y = 1.5 * X + 2.3 * Z + np.random.normal(0, X**2, (num,))

df = pd.DataFrame({'X': X, 'Z': Z, 'Y': Y})
model = smf.ols('Y ~ X + Z', data=df).fit()
se_python = model.bse['X']
print(f"method 1: {se_python:.4f}")

n_sims = 100
x_coefs = []

for i in range(n_sims):
    Z_sim = np.random.normal(0, 1, (num,))
    X_sim = Z_sim + np.random.normal(0, 1, (num,))
    Y_sim = 1.5 * X_sim + 2.3 * Z_sim + np.random.normal(0, X_sim**2, (num,))

    df_sim = pd.DataFrame({'X': X_sim, 'Z': Z_sim, 'Y': Y_sim})
    model_sim = smf.ols('Y ~ X + Z', data=df_sim).fit()
    x_coefs.append(model_sim.params['X'])

se_simulation = np.std(x_coefs, ddof=1)
print(f"method 2: {se_simulation:.4f}")
```
