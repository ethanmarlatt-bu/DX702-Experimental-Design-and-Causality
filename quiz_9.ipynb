### imports

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import skew
```

```python
def simulate(A=1, B=1, C=10, D=1000):
  W = np.random.normal(0,1,D)
  X = W+np.random.normal(0,B,D)
  Y = A*X-W+np.random.normal(0,C,D)
  return Y, X, W
```

### Q1

```python
def get_t_value(A=1, B=1, C=10, D=1000):
    Y, X, W = simulate(A, B, C, D)

    df = pd.DataFrame({'Y': Y, 'X': X, 'W': W})

    model = smf.ols('Y ~ X + W', data=df).fit()

    return model.tvalues['X']

np.random.seed(0)
n_sims = 10000
t_values = np.array([get_t_value(A=1, B=1, C=10, D=1000) for _ in range(n_sims)])

probability = np.mean(np.abs(t_values) > 1.96)
print(f"{probability*100}%")
```

### Q2

```python
def get_estimate(A=1, B=1, C=10, D=1000):
    Y, X, W = simulate(A, B, C, D)

    df = pd.DataFrame({'Y': Y, 'X': X, 'W': W})

    model = smf.ols('Y ~ X + W', data=df).fit()
    
    return model.params['X']

np.random.seed(0)
n_sims = 10000
estimates = np.array([get_estimate(A=1, B=1, C=10, D=1000) for _ in range(n_sims)])

estimate_skew = skew(estimates)
print(f"{estimate_skew:.4f}")
```

### Q3

```python
def power_for_B(B, A=1, C=10, D=1000, n_sims=2000, seed=0):
    rng_state = np.random.get_state()
    np.random.seed(seed)
    t_values = np.array([get_t_value(A=A, B=B, C=C, D=D) for _ in range(n_sims)])
    np.random.set_state(rng_state)
    return np.mean(np.abs(t_values) > 1.96)

candidate_Bs = [0.2, 0.6, 1.8, 5.4]

results = []
for B in candidate_Bs:
    p = power_for_B(B)
    results.append((B, p))

closest = min(results, key=lambda x: abs(x[1] - 0.5))
print(f"{closest[0]}: {closest[1]}")
```

```python
def power_for_A(A, B=1, C=10, D=100, n_sims=2000, seed=0):
    rng_state = np.random.get_state()
    np.random.seed(seed)
    t_values = np.array([get_t_value(A=A, B=B, C=C, D=D) for _ in range(n_sims)])
    np.random.set_state(rng_state)
    return np.mean(np.abs(t_values) > 1.96)

candidate_As = [0.5, 1, 2, 4]

results = []
for A in candidate_As:
    p = power_for_A(A)
    results.append((A, p))

closest = min(results, key=lambda x: abs(x[1] - 0.5))
print(f"{closest[0]}: {closest[1]}")
```
